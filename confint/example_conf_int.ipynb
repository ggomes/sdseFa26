```python
import numpy as np 
import scipy.stats as stats
import matplotlib.pyplot as plt
import pickle
```

## Generate data

```python
np.random.seed(seed=12390103)

# true distribution
muY = 10
sigmaY = 3
Y = stats.norm(loc=muY,scale=sigmaY)

# sample iid data
N = 15
D = Y.rvs(N)

with open('sample_data.pickle','wb') as f:
    pickle.dump(D,f)
```

# Example case 1: $Y$ is Gaussian, $\sigma_Y$ is known.

```python
# load the data
with open('sample_data.pickle','rb') as f:
    D = pickle.load(f)
```

```python
# compute the sample mean
muhat = D.mean()

# radius of the confidence interval
gamma = 0.9
stderr = sigmaY/np.sqrt(N)
factor = abs( stats.norm().ppf((1-gamma)/2) )
rho = stderr*factor

f"Confidence interval: [{muhat-rho:.2f},{muhat+rho:.2f}]"
```
