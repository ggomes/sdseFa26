# Part 1) SciPy

[SciPy](https://docs.scipy.org/doc/scipy/tutorial/index.html#user-guide) provides implementations of many algorithms that are useful to engineers and scientists.  These include algorithms for:

+ numerical differentiation and integration,
+ interpolation,
+ signal processing,
+ linear algebra,
+ optimization,
+ probability and statistics,
+ and many others...

<b>The [stats](https://docs.scipy.org/doc/scipy/tutorial/stats.html) subpackage</b>

SciPy's support for probability and statistics resides in the **`stats`** subpackage. Here you can find implementations of a large number of distribution families, including all of the ones that we cover in this class. 

Below are links to the documentation for the individual distribution classes:

+ Bernoulli: [`stats.bernoulli`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.bernoulli.html#scipy.stats.bernoulli)
+ Binomial: [`stats.binom`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.binom.html#scipy.stats.binom)
+ Exponential: [`stats.expon`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.expon.html#scipy.stats.expon)
+ Continuous uniform: [`stats.uniform`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.uniform.html#scipy.stats.uniform)
+ Discrete uniform: [`stats.randint`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.randint.html#scipy.stats.randint)
+ Gaussian: [`stats.norm`](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.norm.html#scipy.stats.norm)

<br>

<b> Constructors </b>

All of these classes have a common design. You can create an instance of probability distribution by passing the appropriate parameter values to the class constructor.  Here are the parameter values and their translation into our notation. 

| Distribution       | Our notation                | SciPy constructor          | Interpretation of parameters  |
|--------------------|-----------------------------|----------------------------|-------------------------------|
| Bernoulli          | $\mathcal{B}(s)$            | `stats.bernoulli(p)`       | `p`=$s$                       |
| Binomial           | $\mathcal{B}in(n,s)$        | `stats.binom(n,p)`         | `n`=$n$, `p`=$s$              |
| Exponential        | $\mathcal{E}(\lambda)$      | `stats.expon(scale)`       | `scale`=$1/\lambda$           |
| Continuous uniform | $\mathcal{U}^c(a,b)$        | `stats.uniform(loc,scale)` | `loc`=$a$, `scale`=$b-a$      |
| Discrete uniform   | $\mathcal{U}^d(a,b)$        | `stats.randint(low,high)`  | `low`=$a$, `high`=$b+1$       |
| Normal             | $\mathcal{N}(\mu,\sigma^2)$ | `stats.norm(loc,scale)`    | `loc`=$\mu$, `scale`=$\sigma$ |

<br>

<b> Methods </b>

The constructed object then has these methods:

| Method      | Returns                                                             |
|-------------|---------------------------------------------------------------------|
| `pdf(x)`    | Probability density function $p(x)$ (continuous distributions only) |
| `pmf(x)`    | Probability mass function $p(x)$ (discrete distributions only)      |
| `cdf(x)`    | Cumulative distribution function $F(x)$                             |
| `ppf(u)`    | Inverse cdf $F^{-1}(u)$                                             |
| `rvs(size)` | Draw `size` samples.                                                |
| `mean()`    | Mean of the distribution.                                           |
| `var()`     | Variance of the distribution.                                       |
| `std()`     | Standard deviation of the distribution.                             |

# Part 2) Pandas

[Pandas](https://pandas.pydata.org/docs/user_guide/index.html) is an open-source library that specializes in the manipulation and analysis of **tabular data**. 

A **table** is a data type that consists of *columns* and *rows*, as shown in the figure below. 

<center>
<img src="resources/table.png" alt="Alt text" width="700" />
</center>

+ A table has $D$ **columns**. 
    + Each column has a **header**, which is a title that identifies the column
    + Columns are *homeogeneous* in the sense that all of the data contained in a column is of the same type. A column can contain numbers or strings, but not both. 

+ A table has $N$ **rows**. 
    + The first row contains the first element of each column, the second row contains the second element of each column, and so forth. 
    + Each row is identfied by its unique **index** (k1 through kN in the figure). 

In pandas, a table is called a [**DataFrame**](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html), and each column is a [**Series**](https://pandas.pydata.org/docs/reference/api/pandas.Series.html). A DataFrame is a sequence of Series plus an index.

## Part 2.5) Select a row by its index

Row selection in pandas uses the `loc` attribute. `loc` is a dictionary-like structure that is attached to every DataFrame. The set of keys of `loc` is the index of the DataFrame. If you want to select a particular row, you pass the row's index to `loc`. 

For example, to select the row with index 1 and assign it to the variable name `row1` use:

```python 
row1 = T.loc[1]
```
One can also select multiple rows by passing a list of keys to `loc`.

```python 
rows = T.loc[[1,2,3]]
```
`loc` also accepts a second argument for column selection. This can be a single column header or a list of column headers. 

```python 
A = T.loc[[1,2,3],'Cement']
B = T.loc[[1,2,3],['Cement','Blast Furnace Slag']]
```

Finally, one can pass a **boolean mask** to `loc` in order to select rows according to some criterion. A boolean mask is a list (or array, or any sequence) of booleans with as many entries as there are rows in the table. Passing a boolean mask to `loc` selects those rows whose corresponding entry in the mask is `True`. 

We can use boolean masks, for example, to select the concrete mixtures whose compressive strength is greater than 50 MPa. 

```python 
mask = T['Concrete compressive strength']>50   % this produces a Series of booleans
strong_mixtures = T[mask]
```
